```python
from sklearn.datasets import load_diabetes
import numpy as np
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X.shape, y.shape
```

```python
X
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
```

```python
estimators = [('lr',lr), ('dt',dt), ('svr',svr)]
```

```python
for estimator in estimators:
    scores = cross_val_score(estimator[1],X,y,cv=10,scoring='r2')
    print(f"{estimator[0]}: R2 mean={np.mean(scores):.3f}")
```

```python
from sklearn.ensemble import VotingRegressor
vr = VotingRegressor(estimators)
```

```python
scores = cross_val_score(vr,X,y,cv=10,scoring='r2')
print(f"vr: R2 mean={np.mean(scores):.3f}")
```

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vc = VotingRegressor(estimators=estimators, weights=[i,j,k])
            x = cross_val_score(vc,X,y,cv=10,scoring='r2')
            print(f"voting regressors with weights {i},{j},{k}: {np.mean(x)}")
```
