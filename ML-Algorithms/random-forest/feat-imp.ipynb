```python
import pandas as pd 
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('mnist_train.csv')
```

```python
import pandas as pd

# df is your DataFrame

# Generate new pixel names
pixel_names = [f"pixel{i}" for i in range(784)]

# Assign new names (skip the first 'label' column)
df.columns = ['label'] + pixel_names

print(df.columns)
```

```python
df.head()
```

```python
sns.heatmap(df.iloc[0, 1:].values.reshape(28, 28))
```

```python
X = df.drop('label', axis=1).values
y = df['label'].values
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()
rf.fit(X,y)
```

```python
rf.feature_importances_
```

```python
rf.feature_importances_.shape
```

```python
sns.heatmap(rf.feature_importances_.reshape(28,28)) # this tells us which pixels are most important for classification
```
